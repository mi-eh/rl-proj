```python
import pickle
tree_path = "trees/1761721997"
try:
    with open(f"trees/1761721997/policy_forest.pkl", 'rb') as file:
            data = pickle.load(file)
    print("Data loaded successfully:")
    print(data)
except FileNotFoundError:
        print(f"Error: The file '{tree_path}' was not found.")
except Exception as e:
        print(f"An error occurred while loading the pickle file: {e}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

plt.figure(figsize=(40, 30))
plot_tree(data.estimators_[0],filled=True, rounded=True)
plt.title("First Decision Tree in the Random Forest")
plt.savefig("rf_vis.png")
plt.show()
```
